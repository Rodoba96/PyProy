My attempt to scrape my backloggd information

```python
from bs4 import BeautifulSoup
import requests
import pandas as pd
```

```python
url = 'https://backloggd.com/u/SilentStorm63/games/'
page = requests.get(url)
soup = BeautifulSoup(page.text, 'html')
```

```python
print(soup)
```

```python
game_table = soup.find_all('div', class_='justify-content-center')
```

```python
print(game_table)
```

```python
raw_game_titles = game_table[0].find_all('div', class_='game-text-centered')
print(raw_game_titles)

raw_game_url = game_table[0].find_all('a', class_='cover-link')
print(raw_game_url)
```

```python
raw_game_url[1].get('href')
```

```python
game_titles = [ title.text.strip() for title in raw_game_titles ]
print(game_titles)

game_url = [ "https://www.backloggd.com" + url.get('href') for url in raw_game_url ]
print(game_url)
```

```python
print(game_titles[2])
print(game_url[2])
```

```python
print(type(game_titles))
print(type(game_url))
```
