## 1. Importar Librerias

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
import pandas as pd
import time
```

## 2. Nombre de usuario de Backloggd

```python
usuario = "SilentStorm63"
url = f"https://www.backloggd.com/u/{usuario}/games"
```

## 3. Configuracion de Selenium (modo headless)

```python
options = webdriver.ChromeOptions()
options.add_argument("--headless")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")
```

## 4. Iniciar Navegador

```python
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)
driver.get(url)
time.sleep(3)
```

## 5. Scroll para cargar más juegos (ajusta si tienes muchos)

```python
for _ in range(15):
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(1.5)
```

## 6. Extraer elementos

```python
juegos = driver.find_elements(By.CSS_SELECTOR, ".game-text-centered")
links = driver.find_elements(By.CSS_SELECTOR, ".cover-link")
```

## 7. Guardar info

```python
data = []
for juego, link in zip(juegos, links):
    nombre = juego.text.strip()
    href = link.get_attribute("href")
    data.append({
        "Juego": nombre,
        "URL": href
    })

driver.quit()
```

## Crear DataFrame

```python
df = pd.DataFrame(data)
df.to_excel("backloggd_juegos_selenium.xlsx", index=False)
df.head()
```
