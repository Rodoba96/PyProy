```python
# --- 1. Librerías ---
import time
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager

# --- 2. Configuración del navegador ---
options = Options()
options.add_argument("--start-maximized")
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)

# --- 3. Abrir Backloggd ---
usuario = "SilentStorm63"  # tu usuario
pagina = 1
driver.get(f"https://www.backloggd.com/u/{usuario}/games/?page={pagina}")

time.sleep(5)  # Espera que cargue

# --- 4. Extraer juegos de esta página ---
juegos = []

# Ahora buscamos los bloques grandes de cada juego:
bloques_juegos = driver.find_elements(By.CSS_SELECTOR, "div.rating-hover")

print(f"🎮 Juegos encontrados en página {pagina}: {len(bloques_juegos)}")

for bloque in bloques_juegos:
    try:
        # dentro del bloque buscar el link y el nombre
        link_element = bloque.find_element(By.CSS_SELECTOR, "a.cover-link")
        nombre_element = bloque.find_element(By.CSS_SELECTOR, "div.game-text-centered")
        
        url_relativa = link_element.get_attribute("href")
        url = "https://www.backloggd.com" + url_relativa if url_relativa.startswith("/games") else url_relativa
        nombre = nombre_element.text.strip()
        
        juegos.append({"Juego": nombre, "URL": url})
    except Exception as e:
        print(f"⚠️ Error extrayendo datos de un juego: {e}")

# --- 5. Guardar en Excel ---
df = pd.DataFrame(juegos)
df.to_excel("backloggd_juegos_pagina1.xlsx", index=False)

print(f"✅ Archivo generado con {len(juegos)} juegos.")

# --- 6. Cerrar navegador ---
driver.quit()

```

```python
from bs4 import BeautifulSoup
import requests
import pandas as pd
import time

# --- 1. Variables básicas ---
usuario = "SilentStorm63"
base_url = f"https://www.backloggd.com/u/{usuario}/games/?page="
pagina = 1

todos_los_juegos = []
urls_existentes = set()

headers = {
    "User-Agent": "Mozilla/5.0"
}

# --- 2. Loop para recorrer páginas ---
while True:
    print(f"Scrapeando página {pagina}...")
    
    url = base_url + str(pagina)
    response = requests.get(url, headers=headers)
    
    if response.status_code != 200:
        print(f"⚠️ Error al cargar página {pagina}. Código {response.status_code}")
        break
    
    soup = BeautifulSoup(response.content, 'html.parser')
    
    bloques_juegos = soup.find_all('div', class_='rating-hover')

    juegos_nuevos_encontrados = 0  # contador de juegos nuevos

    for bloque in bloques_juegos:
        try:
            link_element = bloque.find('a', class_='cover-link')
            nombre_element = bloque.find('div', class_='game-text-centered')

            if link_element and nombre_element:
                url_relativa = link_element.get('href')
                url_juego = "https://www.backloggd.com" + url_relativa
                nombre_juego = nombre_element.get_text(strip=True)

                if url_juego not in urls_existentes:  # 🚨 solo si es realmente nuevo
                    todos_los_juegos.append({
                        "Juego": nombre_juego,
                        "URL": url_juego
                    })
                    urls_existentes.add(url_juego)
                    juegos_nuevos_encontrados += 1
        except Exception as e:
            print(f"⚠️ Error leyendo un juego: {e}")

    if juegos_nuevos_encontrados == 0:
        print(f"✅ No se encontraron juegos nuevos en página {pagina}. Fin del scraping.")
        break

    pagina += 1
    time.sleep(1)

# --- 3. Guardar en Excel ---
df = pd.DataFrame(todos_los_juegos)
df.to_excel("backloggd_juegos_completo.xlsx", index=False)
df.head(5)

print(f"✅ ¡Scraping completo! Juegos encontrados: {len(df)}")
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import pandas as pd
import time

# --- Mini prueba con 1 juego ---
data_prueba = [
    {"Juego": "Marvel Rivals", "URL": "https://www.backloggd.com/games/marvel-rivals/"}
]
df_prueba = pd.DataFrame(data_prueba)

# --- Configurar navegador ---
options = webdriver.ChromeOptions()
options.add_argument("--headless")
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)

# --- Agregar columnas ---
df_prueba["Horas Jugadas"] = ""
df_prueba["Fecha Inicio"] = ""
df_prueba["Fecha Fin"] = ""

usuario = "SilentStorm63"

for index, row in df_prueba.iterrows():
    url_original = row['URL']
    slug = url_original.split("/games/")[-1].strip("/")
    log_url = f"https://backloggd.com/u/{usuario}/logs/{slug}/"

    print(f"🔍 Visitando: {log_url}")
    driver.get(log_url)
    time.sleep(3)

    # --- Buscar tiempo jugado ---
    try:
        # Buscar el <i> del reloj y luego subir al padre <p> para obtener el texto
        reloj_icono = driver.find_element(By.CLASS_NAME, "fa-stopwatch")
        tiempo_jugado = reloj_icono.find_element(By.XPATH, "./..").text.strip()
    except:
        tiempo_jugado = ""

    # --- Buscar fechas (last played y first played) ---
    try:
        texto_completo = driver.find_element(By.CLASS_NAME, "game-log-view").text
        lineas = texto_completo.split("\n")

        fecha_inicio = ""
        fecha_fin = ""
        for i, linea in enumerate(lineas):
            if "FIRST PLAYED" in linea.upper():
                fecha_inicio = lineas[i+1]
            if "LAST PLAYED" in linea.upper():
                fecha_fin = lineas[i+1]
    except:
        fecha_inicio = ""
        fecha_fin = ""

    # --- Guardar en el DataFrame ---
    df_prueba.at[index, "Horas Jugadas"] = tiempo_jugado
    df_prueba.at[index, "Fecha Inicio"] = fecha_inicio
    df_prueba.at[index, "Fecha Fin"] = fecha_fin

# Mostrar resultados
print("\n✅ Resultado:")
print(df_prueba)

# Cerrar navegador
driver.quit()
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import pandas as pd
import time

# --- CONFIGURACIÓN ---
EXCEL_ENTRADA = "backloggd_juegos_completo.xlsx"
EXCEL_SALIDA = "backloggd_juegos_con_logs.xlsx"
USUARIO = "SilentStorm63"

# --- Cargar Excel existente ---
df = pd.read_excel(EXCEL_ENTRADA)

# Añadir columnas si no existen
for col in ["Horas Jugadas", "Fecha Inicio", "Fecha Fin"]:
    if col not in df.columns:
        df[col] = ""

# --- Configurar navegador ---
options = webdriver.ChromeOptions()
options.add_argument("--headless")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")
options.add_argument('--ignore-certificate-errors')
options.add_argument('--allow-insecure-localhost')
options.add_argument('--disable-web-security')

driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)

# --- Recorrer todos los juegos ---
for index, row in df.iterrows():
    url_original = row['URL']
    juego = row['Juego']
    
    slug = url_original.split("/games/")[-1].strip("/")
    log_url = f"https://backloggd.com/u/{USUARIO}/logs/{slug}/"

    print(f"🔍 {index+1}/{len(df)} - {juego}")
    print(f"    {log_url}")

    try:
        driver.get(log_url)
        time.sleep(2.5)

        # Horas jugadas
        try:
            reloj_icono = driver.find_element(By.CLASS_NAME, "fa-stopwatch")
            horas = reloj_icono.find_element(By.XPATH, "./..").text.strip()
        except:
            horas = ""

        # Fechas
        try:
            texto = driver.find_element(By.CLASS_NAME, "game-log-view").text
            lineas = texto.split("\n")

            fecha_inicio = ""
            fecha_fin = ""
            for i, linea in enumerate(lineas):
                if "FIRST PLAYED" in linea.upper():
                    fecha_inicio = lineas[i+1]
                if "LAST PLAYED" in linea.upper():
                    fecha_fin = lineas[i+1]
        except:
            fecha_inicio = ""
            fecha_fin = ""

        # Guardar en el DataFrame
        df.at[index, "Horas Jugadas"] = horas
        df.at[index, "Fecha Inicio"] = fecha_inicio
        df.at[index, "Fecha Fin"] = fecha_fin

    except Exception as e:
        print(f"⚠️ Error en {log_url}: {e}")

# --- Guardar nuevo Excel ---
df.to_excel(EXCEL_SALIDA, index=False)
print(f"\n✅ Archivo generado: {EXCEL_SALIDA} con {len(df)} juegos.")

# --- Cerrar navegador ---
driver.quit()
```

```python
import pandas as pd
import requests
import time

API_KEY = "ddd05c91688a43bbad088303d18cf599"
ARCHIVO_ENTRADA = "backloggd_juegos_prueba.xlsx"
ARCHIVO_SALIDA = "backloggd_juegos_prueba_enriquecido.xlsx"

df = pd.read_excel(ARCHIVO_ENTRADA)

# Agregar columnas si no existen
for col in ["Nombre RAWG", "Géneros", "Desarrollador", "Plataformas", "Fecha de salida", "Metacritic"]:
    if col not in df.columns:
        df[col] = ""

def obtener_datos_rawg(nombre):
    try:
        url_busqueda = "https://api.rawg.io/api/games"
        params = {"search": nombre, "key": API_KEY, "page_size": 1}
        res = requests.get(url_busqueda, params=params)

        print(f"🔎 Buscando: {nombre}")
        print(f"🛰 Status: {res.status_code}")

        if res.status_code != 200:
            return None

        data = res.json()
        print(f"📦 Respuesta: {data}")

        if not data.get("results"):
            return None

        juego = data["results"][0]
        slug = juego.get("slug", "")

        detalle_url = f"https://api.rawg.io/api/games/{slug}?key={API_KEY}"
        detalle = requests.get(detalle_url).json()

        return {
            "Nombre RAWG": juego.get("name", ""),
            "Géneros": ", ".join([g["name"] for g in juego.get("genres", [])]),
            "Desarrollador": ", ".join([d["name"] for d in detalle.get("developers", [])]) if "developers" in detalle else "",
            "Plataformas": ", ".join([p["platform"]["name"] for p in juego.get("platforms", [])]),
            "Fecha de salida": juego.get("released", ""),
            "Metacritic": juego.get("metacritic", "")
        }
    except Exception as e:
        print(f"⚠️ Error con '{nombre}': {e}")
        return None

# Procesar solo 3 juegos
for index, row in df.iterrows():
    nombre = row["Juego"]
    datos = obtener_datos_rawg(nombre)

    if datos:
        for key, value in datos.items():
            df.at[index, key] = value

    time.sleep(2.5)  # espera para evitar rate limit

df.to_excel(ARCHIVO_SALIDA, index=False)
print(f"\n✅ Archivo generado: {ARCHIVO_SALIDA}")
```
