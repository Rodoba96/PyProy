```python
# --- 1. Librerías ---
import time
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager

# --- 2. Configuración del navegador ---
options = Options()
options.add_argument("--start-maximized")
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)

# --- 3. Abrir Backloggd ---
usuario = "SilentStorm63"  # tu usuario
pagina = 1
driver.get(f"https://www.backloggd.com/u/{usuario}/games/?page={pagina}")

time.sleep(5)  # Espera que cargue

# --- 4. Extraer juegos de esta página ---
juegos = []

# Ahora buscamos los bloques grandes de cada juego:
bloques_juegos = driver.find_elements(By.CSS_SELECTOR, "div.rating-hover")

print(f"🎮 Juegos encontrados en página {pagina}: {len(bloques_juegos)}")

for bloque in bloques_juegos:
    try:
        # dentro del bloque buscar el link y el nombre
        link_element = bloque.find_element(By.CSS_SELECTOR, "a.cover-link")
        nombre_element = bloque.find_element(By.CSS_SELECTOR, "div.game-text-centered")
        
        url_relativa = link_element.get_attribute("href")
        url = "https://www.backloggd.com" + url_relativa if url_relativa.startswith("/games") else url_relativa
        nombre = nombre_element.text.strip()
        
        juegos.append({"Juego": nombre, "URL": url})
    except Exception as e:
        print(f"⚠️ Error extrayendo datos de un juego: {e}")

# --- 5. Guardar en Excel ---
df = pd.DataFrame(juegos)
df.to_excel("backloggd_juegos_pagina1.xlsx", index=False)

print(f"✅ Archivo generado con {len(juegos)} juegos.")

# --- 6. Cerrar navegador ---
driver.quit()

```

# Leer un juego

```python
from bs4 import BeautifulSoup

# Cargar el archivo HTML
with open('backloggd.html', 'r', encoding='utf-8') as file:
    soup = BeautifulSoup(file, 'html.parser')

# Encontrar el primer juego
primer_bloque = soup.find('div', class_='rating-hover')

# Obtener URL
link_element = primer_bloque.find('a', class_='cover-link')
url_relativa = link_element.get('href')
url = "https://www.backloggd.com" + url_relativa

# Obtener Nombre
nombre_element = primer_bloque.find('div', class_='game-text-centered')
nombre = nombre_element.get_text(strip=True)

print("🎮 Nombre:", nombre)
print("🔗 URL:", url)
```
